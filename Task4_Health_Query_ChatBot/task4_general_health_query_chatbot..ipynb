# Task 4: General Health Query Chatbot
**Objective:** Build a chatbot that answers health questions using prompt engineering and Mistral AI.
**Model:** Mistral AI API (mistral-small-latest)
**Approach:** Prompt Engineering with safety filtering

```python
from mistralai.client import Mistral
import warnings
warnings.filterwarnings('ignore')

API_KEY = "5aAtRuu0lKMtSzYs26Mb5SWbXjK2HpfF"
client = Mistral(api_key=API_KEY)
MODEL = "mistral-small-latest"
print("Mistral client initialized successfully!")
```

## 1. System Prompt — Prompt Engineering
Designing the prompt that controls chatbot behavior and safety

```python
SYSTEM_PROMPT = """You are a helpful and friendly medical information assistant.
Your job is to answer general health questions in simple, clear language.

IMPORTANT RULES:
- Always recommend seeing a doctor for serious symptoms
- Never diagnose diseases or prescribe medications
- Keep answers short (3-5 sentences)
- If a question is dangerous or inappropriate, politely decline
- Always end with: 'Please consult a healthcare professional for advice.'
"""
print("System prompt defined!")
```

## 2. Safety Filter
Blocking harmful or dangerous queries before they reach the model

```python
BLOCKED_KEYWORDS = [
    'overdose', 'self-harm', 'suicide',
    'illegal drug', 'kill', 'diagnose'
]

def safety_check(query):
    query_lower = query.lower()
    for keyword in BLOCKED_KEYWORDS:
        if keyword in query_lower:
            return False
    return True

print("Safety filter ready!")
```

## 3. Chatbot Function

```python
def ask_health_chatbot(user_query):
    if not safety_check(user_query):
        return "I cannot assist with that query. Please contact a healthcare professional."
    
    try:
        response = client.chat.complete(
            model=MODEL,
            messages=[
                {"role": "system", "content": SYSTEM_PROMPT},
                {"role": "user", "content": user_query}
            ]
        )
        return response.choices[0].message.content.strip()
    
    except Exception as e:
        return f"Error: {str(e)}"

print("Chatbot function ready!")
```

## 4. Testing the Chatbot

## 5. Interactive Chat Loop

```python
print("Health Query Chatbot — Type 'quit' to exit")
print("=" * 50)

while True:
    user_input = input("You: ").strip()
    if user_input.lower() in ['quit', 'exit', 'bye']:
        print("Bot: Goodbye! Stay healthy!")
        break
    if not user_input:
        continue
    response = ask_health_chatbot(user_input)
    print(f"Bot: {response}")
    print()
```
